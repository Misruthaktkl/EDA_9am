```python
# Step-1: select the continent data
# Step-2: continent data == 'Asia'  it is con1
# Step-3: select the case status data
# step-4: case status data == 'Certified'  is con2
# Step-5: con = con1 & con2
# Step-6: apply main df on con get the len
```

```python
visa_df['continent']
con1 = visa_df['continent']=='Asia'
visa_df['case_status']
con2 = visa_df['case_status']=='Certified'
con = con1 & con2
len(visa_df[con])
```

```python
# from different continents how many ppl are certified


labels=visa_df['continent'].unique()

for i in labels:
    con1=visa_df['continent']==i
    con2=visa_df['case_status']=='Certified'
    con=con1 & con2
    print(len(visa_df[con]))
```

```python
# from different continents how many ppl are certified
labels=visa_df['continent'].unique()

# certified count list=[]
# denied count list=[]
cert_lst=[]
den_lst=[]
for i in labels:
    con1=visa_df['continent']==i
    con2=visa_df['case_status']=='Certified'
    # con3=denied condition
    con3=visa_df['case_status']=='Denied'
    cert_con=con1 & con2
    # den_con= con1 & con3
    den_con= con1 & con3
    cert_lst.append(len(visa_df[cert_con]))
    # append (len(visa_df[den_con]))
    den_lst.append(len(visa_df[den_con]))
df=pd.DataFrame(zip(cert_lst,den_lst),index=labels,columns=['Certified','Denied'])
df
```
