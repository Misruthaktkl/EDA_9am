**Step-13: Drop the column**
- labels: 'IndexLabel | None' = None,
    *,
- axis: 'Axis' = 0,
- index: 'IndexLabel | None' = None,
- columns: 'IndexLabel | None' = None,
- level: 'Level | None' = None,
- inplace: 'bool' = False,

```python
# if i want to drop a index
# I need to use index argument name
df3.drop(index=1)
```

```python
# if i want to drop a column
# I need to use column argument name
df3.drop(columns='Marks')
```

```python
df3.drop(columns=['City','Marks'])
```

```python
df.drop(index=1)
df.drop(columns='Marks')
# ================================
df.drop('Marks') # if i dont give argument, label
df.drop(1)  # 1 become label
# =======================
# we need to give axis
# axis=0 indicates index
# axis=1 indicates columns
```

```python
df3.drop('Marks') 
# by default label='marks'
# if you dont mention axis
# by default axis=0====?
```

```python
df.drop('Marks',axis=1)
```

**inplace**

- when we do the operation the result output is not overwriting

- means the original DataFrame is same as it is.

- in order to do this we need to use **inplace**

- by default inplace= False

- make **inplace=True** if you want to overwrite the result

```python
df.drop('Marks',axis=1,inplace=True)
```
